### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

##*Analisis Extra*

```python
ventas_producto_tienda = (
    tienda_depurada
    .groupby('Producto')
    .agg(
        cantidad_vendida_tienda=('Producto', 'count'),
        precio_tienda=('Precio', 'mean')  # precio promedio por producto
    )
    .reset_index()
)
ventas_producto_tienda.head()
```

```python
ventas_producto_tienda2 = (
    tienda2_depurada
    .groupby('Producto')
    .agg(
        cantidad_vendida_tienda2=('Producto', 'count'),
        precio_tienda2=('Precio', 'mean')  # precio promedio por producto
    )
    .reset_index()
)
ventas_producto_tienda2.head()
```

```python
ventas_producto_tienda3 = (
    tienda3_depurada
    .groupby('Producto')
    .agg(
        cantidad_vendida_tienda3=('Producto', 'count'),
        precio_tienda3=('Precio', 'mean')  # precio promedio por producto
    )
    .reset_index()
)
ventas_producto_tienda3.head()
```

```python
ventas_producto_tienda4 = (
    tienda4_depurada
    .groupby('Producto')
    .agg(
        cantidad_vendida_tienda4=('Producto', 'count'),
        precio_tienda4=('Precio', 'mean')  # precio promedio por producto
    )
    .reset_index()
)
ventas_producto_tienda4.head()
```

Este análisis se realizó para crear un gráfico scatter con el cual comparar el precio vs la cantidad vendida.

##Comparativa entre Precio vs Cantidad vendida

```python
color_base = '#85C1E9'
color_top = '#2E86C1'
color_bottom = '#C0392B'

top_3 = ventas_producto_tienda.nlargest(3, 'cantidad_vendida_tienda')
bottom_3 = ventas_producto_tienda.nsmallest(3, 'cantidad_vendida_tienda')

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# Todos los productos
plt.scatter(
    ventas_producto_tienda['precio_tienda'],
    ventas_producto_tienda['cantidad_vendida_tienda'],
    color=color_base,
    alpha=0.6,
    label='Productos'
)

# Top 3
plt.scatter(
    top_3['precio_tienda'],
    top_3['cantidad_vendida_tienda'],
    color=color_top,
    s=130,
    label='Top 3 más vendidos'
)

# Bottom 3
plt.scatter(
    bottom_3['precio_tienda'],
    bottom_3['cantidad_vendida_tienda'],
    color=color_bottom,
    s=130,
    label='Top 3 menos vendidos'
)

# Etiquetas SOLO para los 6 productos
offsets = [(5, 5), (10, 15), (-15, 10)]

for i, (_, row) in enumerate(top_3.iterrows()):
    plt.annotate(
        row['Producto'],
        (row['precio_tienda'], row['cantidad_vendida_tienda']),
        textcoords="offset points",
        xytext=offsets[i % len(offsets)],
        fontsize=9,
        color=color_top
    )

for _, row in bottom_3.iterrows():
    plt.annotate(
        row['Producto'],
        (row['precio_tienda'], row['cantidad_vendida_tienda']),
        textcoords="offset points",
        xytext=(10, -10),
        fontsize=9,
        color=color_bottom
    )

plt.xlabel('Precio promedio del producto')
plt.ylabel('Cantidad vendida')
plt.title('Relación entre precio promedio y cantidad vendida por producto')
plt.legend()
plt.grid(True)

plt.show()
```
